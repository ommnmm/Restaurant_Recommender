```python
df = pd.read_csv('amazon_reviews_us_Books_v1_02.tsv',sep='\t',engine='python',on_bad_lines='skip')
```

```python
df.shape
df.columns
df.customer_id.value_counts()
df.product_id.value_counts()
```

```python
df.customer_id.value_counts().describe()
s = df.customer_id.value_counts()
s[s>1.].shape
selected = list(s[s>1.].index())
```

```python
selected = list(s[s>100.].index)
df[df.customer_id.isin(selected)].to_csv('selected.csv')
```

```python
df2 = pd.read_csv('selected.csv')
df2.product_id.value_counts().describe()
```

```python
df2.drop(columns=['Unnamed: 0','Unnamed: 0.1'])
```

```python
s = df2.product_id.value_counts()
s[s>2.].shape
selected = list(s[s>2.].index)
df2[df2.product_id.isin(selected)].shape
df2[df2.product_id.isin(selected)].to_csv('selected.csv')
```
